```python
import pandas as pd
import numpy as np

print("--- PROCESSING CMHC RENTAL MARKET DATA (2024 & 2025) ---")
# 1. Load 2024 Met (Barrie CMA) and CSD (Simcoe municipalities)
met_2024 = pd.read_csv('Met_202410_CDP.csv')
csd_2024 = pd.read_csv('CSD_202410_CDP.csv')

# Filter 2024 for Barrie (`METNAME` containing Barrie) and Simcoe municipalities (`CSDNAME` containing Orillia/Midland/Simcoe)
barrie_met_24 = met_2024[met_2024['METNAME'].str.contains('Barrie', case=False, na=False)]
simcoe_csd_24 = csd_2024[csd_2024['CSDNAME'].str.contains('Orillia|Midland|Barrie|Simcoe', case=False, na=False)]

# 2. Load 2025 Excel files
met_2025 = pd.read_excel('Met_202510_CDP.xlsx', sheet_name='Met')
csd_2025 = pd.read_excel('CSD_202510_CDP.xlsx', sheet_name='CSD')

barrie_met_25 = met_2025[met_2025['METNAME'].str.contains('Barrie', case=False, na=False)]
simcoe_csd_25 = csd_2025[csd_2025['CSDNAME'].str.contains('Orillia|Midland|Barrie|Simcoe', case=False, na=False)]

print(f"Loaded CMHC rental rows -> 2024 Barrie Met: {len(barrie_met_24)}, 2025 Barrie Met: {len(barrie_met_25)}")


print("\n--- PROCESSING 2023 WORKING POVERTY (T1FF) DATA ---")
# 3. Load Working Poverty Table
wp_df = pd.read_excel('R25055 - Working Poor, 2023.xlsx', sheet_name='2023 - WP_Table_1')
# Filter for Simcoe County / Barrie / Orillia / Midland
simcoe_wp = wp_df[wp_df.astype(str).apply(lambda x: x.str.contains('Simcoe|Barrie|Orillia|Midland', case=False, na=False)).any(axis=1)]
print(f"Extracted {len(simcoe_wp)} matching rows from the 2023 Working Poverty table.")


print("\n--- PROCESSING ONTARIO MONTHLY CPI INFLATION ---")
# 4. Load CPI Inflation Table
# (Assuming standard StatsCan CSV format with date/value columns)
cpi_df = pd.read_csv('ontario_cpi_monthly.csv', skiprows=9) # adjust skiprows depending on header structure
print("CPI Inflation data loaded successfully.")

# Save out processed intermediate files
simcoe_wp.to_excel('simcoe_working_poverty_clean.xlsx', index=False)
print("\nSuccess! Economic extraction code completed. Ready to merge with daily weather shocks.")
```
